```python
import os, sys, sqlite3
import pandas as pd

sys.path.append("..")   # this notebook sits in notebooks/, so .. is the project folder where src lives
from src.config import PATHS

raw_folder = os.path.join(PATHS["raw_data"], "ds_cqi_jldbc_2018")
print(os.listdir(raw_folder))   # should list arabica_data_cleaned.csv and robusta_data_cleaned.csv

db_path = os.path.join(PATHS["clean_data"], "cqi_project.db")
con = sqlite3.connect(db_path)
pd.read_csv(os.path.join(raw_folder, "arabica_data_cleaned.csv")).to_sql("raw_arabica", con, if_exists="replace", index=False)
pd.read_csv(os.path.join(raw_folder, "robusta_data_cleaned.csv")).to_sql("raw_robusta", con, if_exists="replace", index=False)
con.close()
print("database ready: data/clean/cqi_project.db")
```
